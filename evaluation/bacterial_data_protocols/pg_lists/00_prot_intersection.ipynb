```python
import json
import pandas as pd
import sys
import matplotlib.pyplot as plt

sys.path.append('../../../evaluation_utils/')
from plots import upset_plot
from evaluation.evaluation_func import read_results
```

```python
# Read the JSON file into a Python dictionary
with open('/home/yuliya/repos/cosybio/FedProt/evaluation/bacterial_data_protocols/pg_lists/pg_lists.json', 'r') as file:
    pg_lists = json.load(file)

# transform list into pandas df column
for key in pg_lists.keys():
    pg_lists[key] = pd.DataFrame(pg_lists[key], columns=['proteins'])
```

```python
balanced_plot = upset_plot.generate_upset_plot(
    pg_lists,
    colname='proteins',
    title='Upset plot of PGs\ndifferent preprocessing, bacterial dataset\n',
    categories=pg_lists.keys(),
    save_plot=True, 
    size_fig=(4, 3.5),
    path_svg='/home/yuliya/repos/cosybio/FedProt/evaluation/bacterial_data_protocols/pg_lists/upset_plot_all.svg',
)
```
