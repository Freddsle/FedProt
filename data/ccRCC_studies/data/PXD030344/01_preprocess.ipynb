```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
```

# Read data

```python
# read report in xlsx format
report = pd.read_excel(
   'article/ccRCC proteome quantification matrix.xlsx', 
   sheet_name=2, index_col=1
   )
```

```python
# skip first column
report = report.iloc[:, 1:]
```

```python
# create metadata file based on the report
# colnames of the report put in Sample column
metadata = pd.DataFrame(report.columns, columns=['Sample'])

# if Sample value ends with TA - put Control in Condition column, if with T - put Tumor
metadata['Condition'] = np.where(metadata['Sample'].str.endswith('TA'), 'Normal', 'Tumor')
metadata['Dataset'] = 'PXD030344'
```

```python
# check if non-inque values are in index of report
print(report.index.duplicated().sum())
# print which one
print(report.index[report.index.duplicated()])
```

```python
# replace NA with -1
report.fillna(-1, inplace=True)
# if non-unique values are in Gene, keep only the row with the highest values
report = report.groupby(report.index).max()
# replace -1 with NA
report.replace(-1, np.nan, inplace=True)
```

```python
# save report to csv
report.to_csv('report.csv', sep='\t')
# save metadata to csv
metadata.to_csv('metadata.csv', index=False, sep='\t')
```
